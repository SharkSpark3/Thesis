# Dataset Understanding

Public portfolio edition prepared for GitHub and Databricks. Credentials are read from environment variables; research data and generated artifacts are not committed to Git.


```python
# Databricks uses Unity Catalog Volumes; no Google Drive mount is required.
```

```python
import pandas as pd
```

```python
# Load the dataset
path_gov = "/Volumes/main/default/thesis_project/GovernmentDocument/gov-report-qs/processed_20250804_234209/qs_all_qa_evidence.parquet"
df = pd.read_parquet(path_gov)
```

```python
# Calculate max length (in characters)
df["gold_answer_length"] = df["gold_answer"].astype(str).apply(len)
max_len = df["gold_answer_length"].max()
```

```python
# Get the longest entry
longest_answer = df.loc[df["gold_answer_length"].idxmax(), "gold_answer"]
```

```python
print(f"Max gold_answer length: {max_len} characters")
print("\nExample of the longest answer:\n", longest_answer)
```
